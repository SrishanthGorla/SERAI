```python
import dns.resolver
import pandas as pd
```

```python
import dns.resolver

domains = [
    ("google.com", "A"),
    ("microsoft.com", "A"),
    ("github.com", "AAAA"),
    ("amazon.com", "A"),
    ("google.com", "MX")
]

results = []

for domain, record_type in domains:
    try:
        answers = dns.resolver.resolve(domain, record_type)

        for answer in answers:
            results.append({
                "Domain": domain,
                "Query Type": record_type,
                "Response": str(answer),
                "TTL": answers.rrset.ttl
            })

    except Exception as e:
        results.append({
            "Domain": domain,
            "Query Type": record_type,
            "Response": str(e),
            "TTL": "-"
        })

results
```

```python
df = pd.DataFrame(results)

display(df)
```

```python
print("DNS Query Types:")
print("-----------------")

for record_type in ["A", "AAAA", "MX"]:
    print(f"{record_type} record:")

    for result in results:
        if result["Query Type"] == record_type:
            print("  Domain   :", result["Domain"])
            print("  Response :", result["Response"])
            print("  TTL      :", result["TTL"])

    print()
```

```python
print("DNS Traffic Analysis")
print("====================")

print("1. A records return IPv4 addresses.")
print("2. AAAA records return IPv6 addresses.")
print("3. MX records identify mail servers.")
print("4. TTL indicates how long a DNS response can be cached.")
print("5. Consistent responses from trusted DNS servers generally indicate normal DNS behavior.")
print("6. Unexpected IP addresses, unusual TTL values, or unexpected DNS servers may require investigation.")
```

```python
expected_ip = "142.250.195.14"
observed_ip = "192.0.2.123"

print("DNS Security Check")
print("==================")

print("Expected IP :", expected_ip)
print("Observed IP :", observed_ip)

if expected_ip != observed_ip:
    print("\n⚠️ WARNING: DNS response differs from expected value!")
    print("This could indicate a DNS configuration problem or possible DNS poisoning.")
else:
    print("\n✓ DNS response matches the expected value.")
```

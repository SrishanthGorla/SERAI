```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

fig, ax = plt.subplots(figsize=(15, 3.8))
ax.set_xlim(0, 15)
ax.set_ylim(0, 3)
ax.axis('off')

nodes = [
    ("1. Reconnaissance\n(AML.T0002)\nMap Vector Store", "#37474F"),
    ("2. Supply Chain\n(AML.T0010)\nTaint Tokenizer", "#C62828"),
    ("3. Data Poisoning\n(AML.T0020)\nCorrupt Knowledge", "#D84315"),
    ("4. Prompt Injection\n(AML.T0054)\nExecute Override", "#B71C1C"),
    ("5. Data Leakage\n(AML.T0043)\nExfiltrate PII", "#4A148C")
]

box_width = 2.4
box_height = 1.6

for i, (text, color) in enumerate(nodes):
    x = 0.5 + i * 2.9
    y = 0.7
    rect = patches.FancyBboxPatch(
        (x, y), box_width, box_height,
        boxstyle="round,pad=0.15",
        linewidth=1.5, edgecolor="black", facecolor=color
    )
    ax.add_patch(rect)
    ax.text(
        x + box_width/2, y + box_height/2, text,
        ha="center", va="center", color="white", fontsize=9, fontweight="bold"
    )

    if i < len(nodes) - 1:
        ax.annotate(
            '', xy=(x + box_width + 0.45, y + box_height/2),
            xytext=(x + box_width, y + box_height/2),
            arrowprops=dict(arrowstyle="-|>", lw=2.5, color="#212121", mutation_scale=15)
        )

plt.title("End-to-End Adversarial Attack Path Across RAG System Lifecycle", fontsize=12, fontweight='bold', pad=15)
plt.tight_layout()
plt.savefig("rag_attack_path_diagram.png", dpi=300)
plt.show()
print("Saved diagram as 'rag_attack_path_diagram.png'")
```

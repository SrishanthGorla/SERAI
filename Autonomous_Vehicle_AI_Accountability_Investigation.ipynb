## Cell 6 — Technical analysis: stopping distance and response delay

Simplified stopping distance = reaction distance + braking distance. Reaction distance = speed × response time. Braking distance = speed² / (2 × deceleration). This idealized calculation assumes a level road and constant deceleration; it omits real vehicle dynamics, actuator delays and tire-road variation.

```python
def stopping_distance(speed_kmh, response_time, deceleration):
    speed = speed_kmh / 3.6
    reaction = speed * response_time
    braking = speed**2 / (2 * deceleration)
    return reaction, braking, reaction + braking

rows = []
for speed in [30,40,50,60]:
    for delay in [0.3,1.0,1.5]:
        reaction, braking, total = stopping_distance(speed, delay, 6.0)
        rows.append({"Speed_kmh":speed,"Response_Time_s":delay,
                     "Reaction_Distance_m":round(reaction,2),
                     "Braking_Distance_m":round(braking,2),
                     "Total_Stopping_Distance_m":round(total,2)})
braking_analysis = pd.DataFrame(rows)
print("SIMPLIFIED STOPPING-DISTANCE ANALYSIS")
display(braking_analysis)

plt.figure(figsize=(9,5))
sns.lineplot(data=braking_analysis, x="Speed_kmh", y="Total_Stopping_Distance_m", hue="Response_Time_s", marker="o")
plt.title("Effect of Speed and Response Delay on Stopping Distance")
plt.xlabel("Speed (km/h)")
plt.ylabel("Stopping distance (m)")
plt.tight_layout()
plt.show()
```

## Written conclusion for the assignment

The case study demonstrates that an autonomous vehicle accident can result from multiple interacting failures. Poor lighting and limited training examples may contribute to a perception error; inadequate uncertainty handling, sensor fusion, planning or braking may allow that error to become a collision.

Accountability must be examined across the AI lifecycle. Developers must build robust systems and document limitations; testers must validate difficult scenarios; manufacturers must ensure safe integration and deployment; operators must follow defined supervision procedures; and regulators must establish and enforce appropriate safety requirements.

Preventive measures include diverse training data, sensor redundancy, uncertainty-aware perception, emergency braking, reduced speed in poor visibility, minimum-risk fail-safe behavior, scenario-based testing and continuous monitoring.
